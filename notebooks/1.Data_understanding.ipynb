```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
RAW_PATH = Path("../data/raw")

print(RAW_PATH)
```

```python
customers = pd.read_csv(RAW_PATH / "olist_customers_dataset.csv")
geolocation = pd.read_csv(RAW_PATH / "olist_geolocation_dataset.csv")
order_items = pd.read_csv(RAW_PATH / "olist_order_items_dataset.csv")
payments = pd.read_csv(RAW_PATH / "olist_order_payments_dataset.csv")
reviews = pd.read_csv(RAW_PATH / "olist_order_reviews_dataset.csv")
orders = pd.read_csv(RAW_PATH / "olist_orders_dataset.csv")
products = pd.read_csv(RAW_PATH / "olist_products_dataset.csv")
sellers = pd.read_csv(RAW_PATH / "olist_sellers_dataset.csv")
translation = pd.read_csv(RAW_PATH / "product_category_name_translation.csv")
```

```python
tables = {
    "customers": customers,
    "geolocation": geolocation,
    "order_items": order_items,
    "payments": payments,
    "reviews": reviews,
    "orders": orders,
    "products": products,
    "sellers": sellers,
    "translation": translation
}
```

```python
for name, df in tables.items():
    missing = pd.DataFrame({
        "MISSING_COUNT": df.isna().sum(),
        "MISSING_PERCENT": (df.isna().mean() * 100).round(2)
    })

    missing = missing[missing["MISSING_COUNT"] > 0]

    print("\n" + "=" * 70)
    print(name.upper())
    print("=" * 70)
    print(missing)
```

```python
categorical_columns = {
    "orders": ["order_status"],
    "payments": ["payment_type"],
    "reviews": ["review_score"],
    "products": ["product_category_name"],
    "sellers": ["seller_state"],
    "customers": ["customer_state"]
}
```

```python
for table_name, columns in categorical_columns.items():
    df = tables[table_name]

    print("\n" + "=" * 60)
    print(table_name.upper())
    print("=" * 60)

    for col in columns:
        print(f"\n{col}")
        print(df[col].value_counts(dropna=False).head(20))
```

```python
dictionary_parts = []

for table_name, df in tables.items():
    temp = pd.DataFrame({
        "TABLE": table_name,
        "COLUMN": df.columns,
        "DATATYPE": df.dtypes.astype(str).values,
        "MISSING_COUNT": df.isna().sum().values,
        "MISSING_PERCENT": (
            df.isna().mean().values * 100
        ).round(2)
    })

    dictionary_parts.append(temp)

data_dictionary = pd.concat(
    dictionary_parts,
    ignore_index=True
)

data_dictionary.head()
```
